## Outlier Detection and Treatment

```python
# Visualize outliers using boxplot for key numerical features
plt.figure(figsize=(12,5))
plt.title('Boxplot of Numerical Features')
sns.boxplot(data=df[numerical_cols])
plt.show()
```

```python
# Create a function to remove outliers
# Using IQR Method

def remove_outliers(df, column):
  Q1 = df[column].quantile(0.25)
  Q3 = df[column].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  df_clean = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
  return df_clean
```

```python
# Apply outlier function to columns

for columns in numerical_cols:
  df = remove_outliers(df, columns)
```

```python
# Initialize smote
smote = SMOTE(random_state=42)

# Apply SMOTE to x, y
x_resampled, y_resampled = smote.fit_resample(x, y)

# Check distribution of target variable after resampling
print(y_resampled.value_counts())
```

## Train Test Split

Stratify: A technique used to ensure that the distribution of certain characteristics (typically class labels) within a dataset is maintained across different subsets, particularly when splitting data for training, validation, and testing.

It is to ensure that the proportion of each class in the original dataset is preserved in the resulting subsets (e.g., training set, test set). This is crucial, especially in cases of imbalanced datasets where some classes have significantly fewer samples than others.

In scikit-learn's train_test_split function, the stratify parameter is used to enable stratified splitting. When stratify is set to the target variable (e.g., y), the function ensures that the train and test sets have approximately the same proportion of each class as the original dataset.

```python
# Encode the resampled target variable
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_resampled_encoded = le.fit_transform(y_resampled)

# Split the dataset in train and test sets
x_train, x_test, y_train, y_test = train_test_split(x_resampled, y_resampled_encoded, test_size=0.2, random_state=42, stratify=y_resampled_encoded)
```

## Model Training

```python
# Model Predict Function

def prediction(model):

  # Training
  model.fit(x_train,y_train)

  # Predictions
  y_pred = model.predict(x_test)

  # Evaluation
  acc = accuracy_score(y_test, y_pred)
  cr = classification_report(y_test, y_pred)

  # Print results
  print(f'Accuracy Score: {acc}')
  print(f'Classification Report: \n{cr}')

  ConfusionMatrixDisplay.from_estimator(model, x_test, y_test)
  plt.show()
```

```python
# Model 1
lr = LogisticRegression(max_iter=200)
prediction(lr)
```

```python
# Model 2
dt = DecisionTreeClassifier()
prediction(dt)
```

```python
# Model 3
rf = RandomForestClassifier()
prediction(rf)
```

```python
# Model 4
knn = KNeighborsClassifier()
prediction(knn)
```

```python
# Model 5
nb = GaussianNB()
prediction(nb)
```

```python
# Model 6
lbgm = LGBMClassifier()
prediction(lbgm)
```
