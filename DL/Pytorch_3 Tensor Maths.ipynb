## Tensor Math operations
- Add, Subtract, Multiply, Divide, Remainders, Exponents
- Shorthand and Longhand
- Reassignment

```python
import torch
import numpy as np
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
```

```python
# addition
tensor_a + tensor_b
```

```python
# addition longhand
torch.add(tensor_a, tensor_b)
```

```python
# subtract
tensor_b - tensor_a
```

```python
# subtract longhand
torch.sub(tensor_b, tensor_a)
```

```python
# multiplication
tensor_a * tensor_b
```

```python
# multiplication longhand
torch.mul(tensor_a, tensor_b)
```

```python
torch.multiply(tensor_a,tensor_b)
```

```python
# divide
tensor_b / tensor_a # converts it float
```

```python
# divide by any number
tensor_b / 2
```

```python
# division longhand
torch.div(tensor_b, tensor_a)
```

```python
# remainders using modulus
tensor_b % tensor_a
```

```python
# remainder longhand
torch.remainder(tensor_b, tensor_a)
```

```python
# exponents / power
tensor_a ** tensor_b
```

```python
# exponent / power longhand
torch.pow(tensor_a, tensor_b)
```

```python
# another way to do longhand
tensor_a.add(tensor_b)
tensor_a
```

```python
# reassignment
tensor_a = tensor_a + tensor_b
tensor_a
```

```python
# reassignment longhand
tensor_a = torch.tensor([1,2,3,4])
tensor_a.add_(tensor_b)
tensor_a
```
