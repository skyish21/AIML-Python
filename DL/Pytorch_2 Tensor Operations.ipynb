```python
import torch
```

```python
# create a tensor in a range of 10
my_torch = torch.arange(10)
my_torch
```

```python
# reshape
my_torch = my_torch.reshape(2,5) # reshape should make sense. (2,6) won't
my_torch
```

```python
# reshape if we dont know the number of items using -1
# The special value -1 is used to automatically infer the correct dimension.
my_torch1 = torch.arange(10)
my_torch1
```

```python
# specifying 5 rows, -1 computes number of columns: 10/5 = 2 columns
my_torch1 = my_torch1.reshape(5,-1)
my_torch1
```

```python
# specify 5 columns, -1 computes number of rows: 10/5 = 2 rows
my_torch1 = my_torch1.reshape(-1, 5)
my_torch1
```

```python
my_torch2 = torch.arange(15)
my_torch2
```

```python
# the view function works similarly to reshape
my_torch3 = my_torch2.view(3,5)
my_torch3
```

```python
# slices
my_torch6 = torch.arange(10)
my_torch6
```

```python
# grab a specific item/elemtent within the tensor
my_torch6[7]
```

```python
# grab a slice from the tensor
my_torch7 = my_torch6.reshape(2,5)
my_torch7
```

```python
my_torch7[:,1] # grabs the items at index 1 or column 1
```

```python
my_torch7 = my_torch6.reshape(5,2)
my_torch7
```

```python
my_torch7[:,1] # grabs the items at index 1 or column 1
```

```python
# return a column while retaining structure
my_torch7[:,1:]
```
