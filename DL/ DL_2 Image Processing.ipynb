### Resizing the image using Pillow Library

### Convert RGB Images to Gray Scale Images

```python
# Import OpenCV
import cv2
```

```python
img = cv2.imread('/content/blue-budgie.jpg?s=612x612&w=0&k=20&c=Xz-cvvnfM93ZNIaRfVumhZyI9nO1zzd3Lkno7QrogVU=')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
type(grayscale_img)
```

```python
print(grayscale_img.shape) # Only 2 dimensions and 1 channel
```

```python
# Colab doesn't allow cv2.imshow() function

from google.colab.patches import cv2_imshow
cv2_imshow(grayscale_img)
```
