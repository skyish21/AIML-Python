## Tensors

- A torch.tensor is a multi-dimensional matrix containing elements of a single data type.
- Similar to numpy arrays, but full of fun things that make them work better on GPU's (vs regualer CPU's).
- Default data type is float 32.
- More suitable for deep learning than a numpy array.

```python
import torch
import numpy as np
```

## Numpy Arrays

```python
np1 = np.random.rand(3,4) # 3 rows, 4 columns with random values
np1
```

## Tensors

```python
tensor_2d = torch.randn(3,4) # 3 rows, 4 columns with random values
tensor_2d
```

```python
tensor_3d = torch.zeros(2,3,4) # 2 3x4 tensors with only zeros
tensor_3d
```

```python
tensor_3d.dtype
```

```python
## create tensor from numpy array
my_tensor = torch.tensor(np1)
my_tensor
```
